```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
model = load_model("model.h5")
```

```python
import joblib

scaler = joblib.load("scaler.pkl")

geo_encoder = joblib.load("geo_encoder.pkl")

gender_encoder = joblib.load("gender_encoder.pkl")
```

```python
input_data={
    "CreditScore": [600],
    "Geography": ["France"],
    "Gender": ["Male"],                          
    "Age": [40],
    "Tenure": [3],
    "Balance": [60000],
    "NumOfProducts": [2],
    "HasCrCard": [1],
    "IsActiveMember": [1],
    "EstimatedSalary": [50000]

}
```

```python
input_df = pd.DataFrame(input_data)
```

```python
input_df
```

```python
geo_encoded = geo_encoder.transform(input_df[["Geography"]])
gender_encoded=gender_encoder.transform(input_df[["Gender"]])
```

```python
gender_encoded
geo_encoded
```

```python
geo_encoder.get_feature_names_out(["Geography"])
```

```python
geo_encoded_df=pd.DataFrame(geo_encoded,columns=geo_encoder.get_feature_names_out(["Geography"]))
```

```python
geo_encoded_df
```

```python
input_df=pd.concat([input_df.drop(["Geography"],axis=1),geo_encoded_df],axis=1)
```

```python
input_df
```

```python
input_df.loc[0, "Gender"] =gender_encoded[0]
```

```python
input_df 
```

```python
input_scaled=scaler.transform(input_df)
```

```python
input_scaled
```

```python
# prediction
```

```python
prediction=model.predict(input_scaled)
```

```python
prediction=prediction[0][0]
```

```python
if prediction>0.5:
    print("The customer is likely to leave the bank.")  
else:  
     print("The customer is likely to stay with the bank.")
```
